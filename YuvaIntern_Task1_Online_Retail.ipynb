```python
# Check unusual values in Quantity and UnitPrice

print("Quantity <= 0:", (df["Quantity"] <= 0).sum())
print("UnitPrice <= 0:", (df["UnitPrice"] <= 0).sum())

print("\nLowest quantities:")
display(df.nsmallest(10, "Quantity")[["InvoiceNo", "StockCode", "Quantity"]])

print("\nLowest unit prices:")
display(df.nsmallest(10, "UnitPrice")[["InvoiceNo", "StockCode", "UnitPrice"]])
```

```python
# Examine rows with missing CustomerID
missing_customer = df[df['CustomerID'].isna()]

print("Rows with missing CustomerID:", len(missing_customer))

display(missing_customer.head(20))
```

```python
# Check whether missing CustomerID records have zero prices
print("Missing CustomerID + UnitPrice = 0:",
      ((df['CustomerID'].isna()) & (df['UnitPrice'] == 0)).sum())

# Check whether missing CustomerID records have negative quantities
print("Missing CustomerID + negative Quantity:",
      ((df['CustomerID'].isna()) & (df['Quantity'] < 0)).sum())
```

```python
# Examine duplicate records
duplicates = df[df.duplicated(keep=False)]

print("Total duplicate records:", len(duplicates))

display(duplicates.sort_values(
    by=['InvoiceNo', 'StockCode']
).head(20))
```

```python
# Count exact duplicate rows
exact_duplicates = df[df.duplicated(keep='first')]

print("Exact duplicate rows to remove:", len(exact_duplicates))

# Show a few exact duplicate groups
display(
    df[df.duplicated(keep=False)]
    .sort_values(by=["InvoiceNo", "StockCode", "Quantity"])
    .head(30)
)
```

```python
# Create a copy of the original dataset
clean_df = df.copy()

# Remove exact duplicate rows
clean_df = clean_df.drop_duplicates()

print("Original rows:", len(df))
print("Rows after removing duplicates:", len(clean_df))
print("Rows removed:", len(df) - len(clean_df))
```

```python
# Check negative quantities after removing duplicates
negative_qty = clean_df[clean_df["Quantity"] < 0]

print("Negative quantity records:", len(negative_qty))

# Check how many have cancellation invoice numbers
has_cancel_invoice = negative_qty["InvoiceNo"].astype(str).str.startswith("C")

print("\nNegative quantity records with cancellation invoices:")
print(has_cancel_invoice.value_counts())
```

```python
# Negative quantities without a cancellation invoice
unusual_negative = clean_df[
    (clean_df["Quantity"] < 0) &
    (~clean_df["InvoiceNo"].astype(str).str.startswith("C"))
]

print("Negative quantity records without cancellation invoice:",
      len(unusual_negative))

display(unusual_negative.head(20))
```

```python
# Check whether all unusual negative records have zero UnitPrice
print("UnitPrice = 0:",
      (unusual_negative["UnitPrice"] == 0).sum())

print("UnitPrice > 0:",
      (unusual_negative["UnitPrice"] > 0).sum())

print("UnitPrice < 0:",
      (unusual_negative["UnitPrice"] < 0).sum())

print("\nMissing Description:",
      unusual_negative["Description"].isna().sum())

print("Missing CustomerID:",
      unusual_negative["CustomerID"].isna().sum())
```

```python
# Remove negative-quantity records that are not cancellation invoices
clean_df = clean_df[
    ~(
        (clean_df["Quantity"] < 0) &
        (~clean_df["InvoiceNo"].astype(str).str.startswith("C"))
    )
].copy()

print("Rows after removing non-standard negative records:", len(clean_df))
```

```python
# Check zero UnitPrice records in the current cleaned dataset
zero_price = clean_df[clean_df["UnitPrice"] == 0]

print("Zero UnitPrice records:", len(zero_price))

print("\nMissing Description:",
      zero_price["Description"].isna().sum())

print("Missing CustomerID:",
      zero_price["CustomerID"].isna().sum())

print("\nQuantity <= 0:",
      (zero_price["Quantity"] <= 0).sum())

print("Quantity > 0:",
      (zero_price["Quantity"] > 0).sum())
```

```python
# Inspect descriptions of zero-price records
zero_price_desc = (
    zero_price["Description"]
    .value_counts(dropna=False)
)

print("Number of unique descriptions:", zero_price_desc.shape[0])

display(zero_price_desc.head(30))
```

```python
# Check negative UnitPrice records
negative_price = clean_df[clean_df["UnitPrice"] < 0]

print("Negative UnitPrice records:", len(negative_price))

display(negative_price)
```

```python
# Remove negative UnitPrice accounting adjustment records
clean_df = clean_df[clean_df["UnitPrice"] >= 0].copy()

print("Rows after removing negative-price adjustments:", len(clean_df))
```

```python
# Check missing values in the current cleaned dataset
missing_values = clean_df.isnull().sum()

missing_percent = (missing_values / len(clean_df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": missing_percent.round(2)
})

display(missing_summary)
```

```python
# Calculate IQR-based outlier bounds for Quantity and UnitPrice

Q1_quantity = clean_df["Quantity"].quantile(0.25)
Q3_quantity = clean_df["Quantity"].quantile(0.75)
IQR_quantity = Q3_quantity - Q1_quantity

lower_quantity = Q1_quantity - 1.5 * IQR_quantity
upper_quantity = Q3_quantity + 1.5 * IQR_quantity

Q1_price = clean_df["UnitPrice"].quantile(0.25)
Q3_price = clean_df["UnitPrice"].quantile(0.75)
IQR_price = Q3_price - Q1_price

lower_price = Q1_price - 1.5 * IQR_price
upper_price = Q3_price + 1.5 * IQR_price

print("Quantity IQR bounds:", lower_quantity, "to", upper_quantity)
print("UnitPrice IQR bounds:", lower_price, "to", upper_price)

print("\nQuantity outliers:",
      ((clean_df["Quantity"] < lower_quantity) |
       (clean_df["Quantity"] > upper_quantity)).sum())

print("UnitPrice outliers:",
      ((clean_df["UnitPrice"] < lower_price) |
       (clean_df["UnitPrice"] > upper_price)).sum())
```

```python
# Inspect the largest Quantity values
print("Top 20 highest Quantity values:")
display(
    clean_df.nlargest(20, "Quantity")[
        ["InvoiceNo", "StockCode", "Description", "Quantity", "UnitPrice", "CustomerID", "Country"]
    ]
)

# Inspect the highest UnitPrice values
print("\nTop 20 highest UnitPrice values:")
display(
    clean_df.nlargest(20, "UnitPrice")[
        ["InvoiceNo", "StockCode", "Description", "Quantity", "UnitPrice", "CustomerID", "Country"]
    ]
)
```

```python
# Check unusual/non-product descriptions among high UnitPrice records

unusual_keywords = [
    "AMAZON",
    "POSTAGE",
    "MANUAL",
    "ADJUST",
    "BAD DEBT",
    "FEE"
]

pattern = "|".join(unusual_keywords)

unusual_records = clean_df[
    clean_df["Description"].astype(str).str.upper().str.contains(
        pattern, na=False
    )
]

print("Number of unusual/non-product records:", len(unusual_records))

display(
    unusual_records[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "CustomerID", "Country"]
    ].sort_values("UnitPrice", ascending=False).head(30)
)
```

```python
# Count the unusual/non-product records by description

print(
    unusual_records["Description"]
    .value_counts(dropna=False)
    .head(30)
)
```

```python
# Count clearly non-product transaction types

non_product = clean_df[
    clean_df["Description"].astype(str).str.upper().isin([
        "POSTAGE",
        "DOTCOM POSTAGE",
        "MANUAL",
        "AMAZON FEE",
        "ADJUSTMENT",
        "ADJUST BAD DEBT"
    ])
]

print("Clearly non-product records:", len(non_product))

print("\nBreakdown:")
print(non_product["Description"].value_counts())
```

```python
# Remove clearly non-product transaction records

non_product_descriptions = [
    "POSTAGE",
    "DOTCOM POSTAGE",
    "MANUAL",
    "AMAZON FEE",
    "ADJUSTMENT",
    "ADJUST BAD DEBT"
]

clean_df = clean_df[
    ~clean_df["Description"].astype(str).str.upper().isin(non_product_descriptions)
].copy()

print("Rows after removing non-product transactions:", len(clean_df))
```

```python
# Remove records with missing product descriptions

clean_df = clean_df.dropna(subset=["Description"]).copy()

print("Final cleaned dataset rows:", len(clean_df))
print("Final cleaned dataset columns:", len(clean_df.columns))

print("\nRemaining missing values:")
print(clean_df.isnull().sum())
```

```python
# Save the final cleaned dataset
output_file = "/content/Online_Retail_Cleaned.csv"

clean_df.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print("File:", output_file)
print("Rows:", len(clean_df))
print("Columns:", len(clean_df.columns))
```
